```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data=pd.read_csv("/content/archive (11).zip",compression="zip")
print(data.shape)
# print(data.describe)
# print(data.info)
print(data.head())
```

```python
print(data.dtypes)
print(data.isnull().sum())
```

```python
dtypes_objects=data.select_dtypes(include= ["object"]).columns
data[dtypes_objects]=data[dtypes_objects].fillna("Unknown")

print(data.isnull().sum())
```

```python
print(data["Unnamed: 0"] )
```

```python
data["Unnamed: 0"]=data["Unnamed: 0" ].interpolate(method="linear")
```

```python
print(data["Price"])
```

```python
price_mean = data["Price"].mean()
data["Price"] = data["Price"].fillna(price_mean)
print(data.isnull().sum())
```
